# Dataset Split for Hyperparameter Tuning
Split the dataset into 70% training and 30% testing sets

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load the dataset
df = pd.read_parquet('clean.parquet')

print("Dataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst few rows:")
print(df.head())
```

```python
# Separate features and target
y = df['trip_duration'].values
X = df.drop('trip_duration', axis=1)

print(f"Features shape: {X.shape}")
print(f"Target shape: {y.shape}")
print(f"\nFeature columns: {X.columns.tolist()}")
```

```python
# Split into 70% training and 30% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.30, 
    random_state=42
)

print("="*60)
print("Dataset Split Complete")
print("="*60)
print(f"Training set: {X_train.shape[0]:,} samples ({X_train.shape[0]/len(X)*100:.1f}%)")
print(f"Test set:     {X_test.shape[0]:,} samples ({X_test.shape[0]/len(X)*100:.1f}%)")
print("="*60)
```

```python
# Save the splits to parquet files for use in hyperparameter tuning
train_df = X_train.copy()
train_df['trip_duration'] = y_train

test_df = X_test.copy()
test_df['trip_duration'] = y_test

train_df.to_parquet('train_70.parquet', index=False)
test_df.to_parquet('test_30.parquet', index=False)

print("✅ Saved train_70.parquet")
print("✅ Saved test_30.parquet")
```

```python
# Verify the saved files
train_loaded = pd.read_parquet('train_70.parquet')
test_loaded = pd.read_parquet('test_30.parquet')

print("\nVerification:")
print(f"Train file shape: {train_loaded.shape}")
print(f"Test file shape: {test_loaded.shape}")
print(f"\nTrain target statistics:")
print(train_loaded['trip_duration'].describe())
print(f"\nTest target statistics:")
print(test_loaded['trip_duration'].describe())
```
